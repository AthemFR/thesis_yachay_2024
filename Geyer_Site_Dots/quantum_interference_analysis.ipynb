```python
#POLARIZATION p_L = p_R = 0
p_L = 1
p_R = 0

#Transmission axis

transmission_u_L_1 = fy.Transmission(energy_axis, gamma, Hamiltonian, site_energies, p_L, p_R)[0]
transmission_d_L_1 = fy.Transmission(energy_axis, gamma, Hamiltonian, site_energies, p_L, p_R)[1]
```

```python
#POLARIZATION UP-UP
p_L_1 = -1
p_R_1 = 0

#Transmission axis

transmission_u_1_L_1 = fy.Transmission(energy_axis, gamma, Hamiltonian, site_energies, p_L_1, p_R_1)[0]
transmission_d_1_L_1 = fy.Transmission(energy_axis, gamma, Hamiltonian, site_energies, p_L_1, p_R_1)[1]
```

```python
T_UP_L_1 = transmission_u_L_1 + transmission_d_L_1
T_DOWN_L_1 = transmission_u_1_L_1 + transmission_d_1_L_1

SP1_L_1 = ( (T_UP_L_1-T_DOWN_L_1)/(T_UP_L_1+T_DOWN_L_1) )*100
```

```python
# Crear la figura y los ejes
fig, ax1 = plt.subplots(figsize=(14, 7), dpi=300)

# Define the first axis and its y-scale (linear scale)
ax1.plot(energy_axis, SP1_L_1, color="#DC143C", label=r"$SP$", lw=4)
ax1.set_xlabel(r"$E\,[meV]$", fontsize=30)
ax1.set_ylabel(r"$SP\,[\%]$", fontsize=30)

# Make the second y axis (logarítmic scale)
ax2 = ax1.twinx()
ax2.plot(energy_axis, transmission_u_L_1 + transmission_d_L_1, "#000000", linestyle="-.", label=r"$T_{UP}$", lw=2)
ax2.plot(energy_axis, transmission_u_1_L_1 + transmission_d_1_L_1, "#1f77b4", linestyle="-.", label=r"$T_{DOWN}$", lw=2)

ax2.set_ylabel(r"$Transmission_{total}$", fontsize=30)

# Obtain legends and lines for both axes
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()

# Locate the legends such that it is not superposed
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper right', fontsize=20)

# Adjust the x axis ticks
ax1.set_xlim(-6, 6)

# Establish the x axis ticks in intervals of 1
ax1.set_xticks(range(-5, 4 + 2, 1))

# Set the fontsize for the x and y ticks
ax1.tick_params(axis='x', labelsize=30)
ax1.tick_params(axis='y', labelsize=30)
ax2.tick_params(axis='y', labelsize=30)

# Headline
plt.suptitle("Spin Polarization " + r"($t=$" + str(t/gamma) + r"$\gamma$, " + r"$\omega=$" + str(om/gamma) + r"$\gamma\,\,&\,B=$" + str(B) + r"$\lambda_{2}\,meV$)", fontsize=30, y=0.95)

# Show the image
plt.show()

# Save the image
fig.savefig('spin_polarization_helix_site_dots.jpg', bbox_inches='tight')
```
