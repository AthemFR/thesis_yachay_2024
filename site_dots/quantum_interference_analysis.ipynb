```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
import functions_spin_p as fy
from importlib import reload

reload(fy)
```

```python
#Energy axis

energy_axis = np.arange(-8, 8 + 0.01, 0.01) #meV
print("Energy axis is", energy_axis,"meV")

#Gamma i.e. width parameter
gamma = 0.5 #meV
print("The width is:",gamma,"meV")

#Site energies:
e_01 = 0 #meV
e_02 = 0 #meV
e_11 = 1 #meV
e_12 = 1 #meV

# Interdot Hopping
v = 4*gamma
```

```python
# do not forget constrained_layout=True to have some space between axes
fig = plt.figure(constrained_layout=True, figsize=(18,11), dpi=300)

#Set the grid for the subplots
gs = GridSpec(1, 1, figure=fig)

#Add the subplots
ax1 = fig.add_subplot(gs[0, 0])

#POLARIZATION = 0
# Transmission spin up Plot
p=0

ax1.grid(linestyle=':', color='grey')
################################################################################################################
#Definition of the Hamiltonian 

#Innerdot Hopping parameter:
om = 0*gamma #meV

#Renormalized site energies array:
e_ren_1_u = np.array([e_01,e_11, om])
e_ren_1_d = np.array([e_01,e_11, om])

e_ren_2_u = np.array([e_02, e_12, om])
e_ren_2_d = np.array([e_02, e_12, om])


site_energies = np.array([e_ren_1_u, e_ren_1_d , e_ren_2_u , e_ren_2_d], dtype = complex)
#print(site_energies)

Hamiltonian = np.array([[0,0,v,0],[0,0,0,v],[v,0,0,0],[0,v,0,0]], dtype = complex)

ax1.plot(energy_axis, fy.Transmission(energy_axis, gamma, Hamiltonian, site_energies, p, p)[0]\
     +fy.Transmission(energy_axis, gamma, Hamiltonian, site_energies, p, p)[1], "b", label=r"$\omega\,=\,$"+str(om/gamma)+r"$\gamma$", linestyle="-.", lw=4)

##################################################################################################################
#Definition of the Hamiltonian 

#Innerdot Hopping parameter:
om = 1*gamma #meV

#Renormalized site energies array:
e_ren_1_u = np.array([e_01,e_11, om])
e_ren_1_d = np.array([e_01,e_11, om])

e_ren_2_u = np.array([e_02, e_12, om])
e_ren_2_d = np.array([e_02, e_12, om])


site_energies = np.array([e_ren_1_u, e_ren_1_d , e_ren_2_u , e_ren_2_d], dtype = complex)
#print(site_energies)

Hamiltonian = np.array([[0,0,v,0],[0,0,0,v],[v,0,0,0],[0,v,0,0]], dtype = complex)

ax1.plot(energy_axis, fy.Transmission(energy_axis, gamma, Hamiltonian, site_energies, p, p)[0]\
     +fy.Transmission(energy_axis, gamma, Hamiltonian, site_energies, p, p)[1], "m", label=r"$\omega\,=\,$"+str(om/gamma)+r"$\gamma$", linestyle="-", lw=4)

#####################################################################################################################
#Definition of the Hamiltonian 

#Innerdot Hopping parameter:
om = 8*gamma #meV

#Renormalized site energies array:
e_ren_1_u = np.array([e_01,e_11, om])
e_ren_1_d = np.array([e_01,e_11, om])

e_ren_2_u = np.array([e_02, e_12, om])
e_ren_2_d = np.array([e_02, e_12, om])


site_energies = np.array([e_ren_1_u, e_ren_1_d , e_ren_2_u , e_ren_2_d], dtype = complex)
#print(site_energies)

Hamiltonian = np.array([[0,0,v,0],[0,0,0,v],[v,0,0,0],[0,v,0,0]], dtype = complex)

ax1.plot(energy_axis, fy.Transmission(energy_axis, gamma, Hamiltonian, site_energies, p, p)[0]\
     +fy.Transmission(energy_axis, gamma, Hamiltonian, site_energies, p, p)[1], "r", label=r"$\omega\,=\,$"+str(om/gamma)+r"$\gamma$", linestyle="--", lw=4)

############################################################################################################################
ax1.set_xlabel(r'$E\,[meV]$',fontsize=30)
ax1.set_ylabel(r'$T(E) = T_{\uparrow}(E)+T_{\downarrow}(E)$',fontsize=30)
ax1.legend(
    fontsize=25,
    title="Intra-hopping "+r"$\omega$",
    title_fontsize=25,
    loc="upper right"
)

# Stablish limits for the subplots
ax1.set_ylim(0, 2.5)  
ax1.set_xlim(-7, 7)  

# Set the fontsize for the x and y ticks
ax1.tick_params(axis='x', labelsize=30)
ax1.tick_params(axis='y', labelsize=30)

# Ajustar el diseño y mostrar los subplots
plt.suptitle("Transmission vs Energy "+r" ($p_{L}=p_{R}=$"+str(p)+r" & $V=$"+str(v/gamma)+r"$\gamma$)", fontsize=30)
plt.tight_layout()
plt.show()
fig.savefig('site_dots_Transmission_reference.jpg')
```
