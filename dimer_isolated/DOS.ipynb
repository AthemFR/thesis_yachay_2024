## Plot Density of States

```python
import numpy as np
import matplotlib.pyplot as plt

import sympy as sp
from sympy import exp
```

```python
#Define the omega axis

omega_axis = np.arange(0, 10 + 0.01, 0.01 )
#print(omega_axis)

#Define the parameters

#eta = 0.01
e_1 = 3
e_2 = 4
v = 2
```

```python
#Define the Density of States:

def DOS(omega, eta):
    result = ( 2*(eta**3) + 2*(omega**2)*eta - 2*omega*eta*(e_1 + e_2)\
            + (e_1**2 + e_2**2)*eta + 2*(v**2)*eta)/( (e_1*e_2 - e_1*omega - e_2*omega\
            +omega**2 - eta**2 - v**2)**2 + (eta**2)*(2*omega-e_1-e_2)**2 )
                                            
    return result
```

```python
print(omega_axis)
print(DOS(omega_axis, 0.01))
```

```python
plt.figure(figsize = (8,6))

plt.plot(omega_axis, DOS(omega_axis, 0.01), "g", label = "$\eta$ = 0.01")
#plt.plot(omega_axis, DOS(omega_axis, 0.02), "#F39C12", label = "$\eta$ = 0.04")

plt.xlabel('$\omega$')
plt.ylabel('DOS($\omega$)')

plt.xlim(-0.1,10.1)
plt.ylim(-0.1,100)

plt.title("DOS vs. Frequency")
plt.legend()
plt.show()
```
