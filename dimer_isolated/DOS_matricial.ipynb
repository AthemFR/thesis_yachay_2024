## Plot Density of States: isolated dimer

```python
import numpy as np
import matplotlib.pyplot as plt

import functions_spin_p as function
from importlib import reload
reload(function)
```

```python
#Define the Energy axis

energy_axis = np.arange(0, 10 + 0.01, 0.01 )
#print(omega_axis)

#Define the parameters

eta = 0.01
e_1 = 3
e_2 = 4
v = 2
```

```python
Hamiltonian = np.array([[e_1, v],[v, e_2]], dtype = complex)
print(Hamiltonian)
```

```python
function.DOS(energy_axis, eta, Hamiltonian)
```

```python
print(energy_axis)
print(function.DOS(energy_axis, eta, Hamiltonian))
```

```python
plt.figure(figsize = (8,6))

plt.plot(energy_axis, function.DOS(energy_axis, eta, Hamiltonian), "g", label = "$\eta$ = "+str(eta))
#plt.plot(omega_axis, DOS(omega_axis, 0.02), "#F39C12", label = "$\eta$ = 0.04")

plt.xlabel('$\omega$')
plt.ylabel('DOS($\omega$)')

plt.xlim(-0.1,10.1)
plt.ylim(-0.1,100)

plt.title("DOS vs. Frequency")
plt.legend()
plt.show()
```

```python
plt.figure(figsize = (8,6))
plt.grid(linestyle=':', color='grey')

plt.plot(energy_axis, function.DOS(energy_axis, 0.01, Hamiltonian), "g", label = "$\eta$ = 0.01")
plt.plot(energy_axis, function.DOS(energy_axis, 0.02, Hamiltonian), "#F39C12", label = "$\eta$ = 0.02")
plt.plot(energy_axis, function.DOS(energy_axis, 0.04, Hamiltonian), "red", label = "$\eta$ = 0.04")

plt.xlabel('$\omega$')
plt.ylabel('DOS($\omega$)')

plt.xlim(-0.1,10.1)
plt.ylim(-0.1,100)

plt.title("DOS vs. Frequency")
plt.legend()
# Save the image
plt.savefig('DOS_isolated dimer.jpg') #, bbox_inches='tight')
plt.show()

```
