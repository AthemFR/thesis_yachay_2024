```python
#Energy axis

energy_axis = np.arange(-6, 6 + 0.01, 0.01) #meV
print("Energy axis is", energy_axis,"meV")

#Gamma i.e. width parameter
gamma = 0.5 #meV
print("The width is:",gamma,"meV")

#Site energies:
e_1 = 0 #meV
e_2 = 0 #meV

#Hopping parameter:
t = 7*gamma #meV

# Helix geometric parameters:

lam_2 = 0.5 #meV

B = - 2
```

```python
#POLARIZATION UP-NM (Non magnetic)
p_L = 1
p_R = 0

#Transmission axis
L=1
Hamiltonian = np.array([[e_1_u + lam_2*L*B, -(1j)*lam_2*L*B,-t,0],[(1j)*lam_2*L*B, e_1_d-lam_2*L*B, 0,-t],[-t,0, e_2_u + lam_2*L*B, -(1j)*lam_2*L*B],\
                        [0,-t, (1j)*lam_2*L*B, e_2_d - lam_2*L*B]], dtype = complex)
transmission_u1_L_1 = fy.Transmission(energy_axis, gamma, Hamiltonian, p_L, p_R)[0]

L=1
Hamiltonian = np.array([[e_1_u + lam_2*L*B, -(1j)*lam_2*L*B,-t,0],[(1j)*lam_2*L*B, e_1_d-lam_2*L*B, 0,-t],[-t,0, e_2_u + lam_2*L*B, -(1j)*lam_2*L*B],\
                        [0,-t, (1j)*lam_2*L*B, e_2_d - lam_2*L*B]], dtype = complex)
transmission_d1_L_1 = fy.Transmission(energy_axis, gamma, Hamiltonian, p_L, p_R)[1]
```

```python
#POLARIZATION DOWN-NM
p_L_1 = -1
p_R_1 = 0

#Transmission axis
L=1
Hamiltonian = np.array([[e_1_u + lam_2*L*B, -(1j)*lam_2*L*B,-t,0],[(1j)*lam_2*L*B, e_1_d-lam_2*L*B, 0,-t],[-t,0, e_2_u + lam_2*L*B, -(1j)*lam_2*L*B],\
                        [0,-t, (1j)*lam_2*L*B, e_2_d - lam_2*L*B]], dtype = complex)
transmission_u2_L_1 = fy.Transmission(energy_axis, gamma, Hamiltonian, p_L_1, p_R_1)[0]

L=1
Hamiltonian = np.array([[e_1_u + lam_2*L*B, -(1j)*lam_2*L*B,-t,0],[(1j)*lam_2*L*B, e_1_d-lam_2*L*B, 0,-t],[-t,0, e_2_u + lam_2*L*B, -(1j)*lam_2*L*B],\
                        [0,-t, (1j)*lam_2*L*B, e_2_d - lam_2*L*B]], dtype = complex)
transmission_d2_L_1 = fy.Transmission(energy_axis, gamma, Hamiltonian, p_L_1, p_R_1)[1]
```

```python
T_UP_L_1 = transmission_u1_L_1 + transmission_d1_L_1
T_DOWN_L_1 = transmission_u2_L_1 + transmission_d2_L_1

SP1_L_1 = ( (T_UP_L_1-T_DOWN_L_1)/(T_UP_L_1+T_DOWN_L_1) )*100
```

```python
# Crear la figura y los ejes
fig, ax1 = plt.subplots(figsize=(14, 7), dpi=300)

# Define the first axis and its y-scale (linear scale)
ax1.plot(energy_axis, SP1_L_1, color="#DC143C", label=r"$SP$", lw=4)
ax1.set_xlabel(r"$E\,[meV]$", fontsize=30)
ax1.set_ylabel(r"$SP\,[\%]$", fontsize=30)

# Make the second y axis (logarítmic scale)
ax2 = ax1.twinx()
ax2.plot(energy_axis, transmission_u1_L_1 + transmission_d1_L_1, "#000000", linestyle="-.", label=r"$T_{UP}$", lw=2)
ax2.plot(energy_axis, transmission_u2_L_1 + transmission_d2_L_1, "#1f77b4", linestyle="-.", label=r"$T_{DOWN}$", lw=2)

ax2.set_ylabel(r"$Transmission_{total}$", fontsize=30)

# Obtain legends and lines for both axes
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()

# Locate the legends such that it is not superposed
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper center', fontsize=20)

# Adjust the x axis ticks
ax1.set_xlim(-6, 6)
#ax1.set_ylim(-70, 70)

# Establish the x axis ticks in intervals of 1
ax1.set_xticks(range(-6, 5 + 2, 1))
ax1.set_yticks(range(-50, 60 , 50))

# Set the fontsize for the x and y ticks
ax1.tick_params(axis='x', labelsize=30)
ax1.tick_params(axis='y', labelsize=30)
ax2.tick_params(axis='y', labelsize=30)

# Headline
plt.suptitle("Spin Polarization" + r"($t=$" + str(t/gamma) + r"$\gamma\,\,&\,B=$" + str(B) + r"$\lambda_{2}\,meV$)", fontsize=30, y=0.95)

# Adjust layout to prevent cutting off elements
plt.tight_layout(rect=[0, 0, 1, 0.95])

# Show the image
plt.show()

# Save the image
fig.savefig('spin_polarization_helix.jpg', bbox_inches='tight')
```
